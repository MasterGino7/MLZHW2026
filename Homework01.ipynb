```python
AsiaCars = df[df['origin']=='Asia'][['model_year','vehicle_weight']].reset_index(drop=True)
AsiaCars.head(7)
X = AsiaCars.head(7).to_numpy()
XTX = np.dot(X.T, X)
XTXInv = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = np.dot(np.dot(XTXInv,X.T),y)
sum(w)
```
